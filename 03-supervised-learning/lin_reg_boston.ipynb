```python
from sklearn import linear_model
from sklearn.metrics import mean_squared_error, r2_score
```

```python
model = linear_model.LinearRegression()
```

```python
X = housing.drop(['medv'], axis=1)
Y = housing.medv
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, train_size=0.8,random_state=42)
model.fit(X_train, Y_train)
```

```python
Y_pred=model.predict(X_test)
```

```python
print("coefficient is: " ,model.coef_)
print("intercept is: " ,model.intercept_)
print("mean_squared_error is: %2f" %mean_squared_error(Y_test,Y_pred))
print("coefficient of determinant: %2f" %r2_score(Y_test,Y_pred))
```

```python
r2_score(Y_test,Y_pred)
```

```python
from sklearn.metrics import mean_absolute_error

mean_absolute_error(Y_test, Y_pred)

#it tells us on an avg my pridictions is how much units away from true value ie 3.14 here
```
